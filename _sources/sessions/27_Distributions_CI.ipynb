## Intro to (Probability) Distributions

<div class="group">
 <div class="text">  

* **Distribution**: Shows how values in a dataset are spread out. It tells us how often each value occurs.
* **Probability Distribution**: A mathematical function that describes the **likelihood** of different outcomes.
* **Types of Distributions**: 
  + **Discrete**: Used for countable outcomes (e.g., number of students in a class).
  + **Continuous**: Used for measurable outcomes (e.g., height, weight, temperature).
 </div>
 <div class="images">   
    <img src="../images/distributions/distributions.png" alt="distributions">
 </div>

</div>


### Probability Mass Function (PMF)

<div class="group">
 <div class="text">  

* It gives the probability of a specific outcome for **discrete random variables**.
* It is defined as:
  $$ P(X = x) = p(x) $$
  where $p(x)$ is the PMF of the random variable $X$.

* **Example**: If you roll a die 🎲, the **PMF** gives the probability of rolling each number (1 to 6).
* **PMF** is often represented as a bar chart.
 </div>
 <div class="images">   
    <img src="../images/distributions/die_outcome.png" alt="distributions">
 </div>
 </div>


### Probability Density Function (PDF)
<div class="group">
 <div class="text">

* It describes the density of the probabilty for **continuous random variables**.
* **Probaility** is represented as an **area under the curve over a range of values**.
* It is defined as:
  $$ P(a < X < b) = \int_{a}^{b} f(x) \, dx $$ 
  where $f(x)$ is the PDF of the random variable $X$.

* **Example**: What is the probability that a randomly chosen person's height is between 160 cm and 180 cm?

  </div>
  <div class="images">
      <img src="../images/distributions/standard_distr.png" alt="Probability Density Function">

### Cumulative Distribution Function (CDF)
<div class="group">
 <div class="text"> 

+ It gives the cumulative probability that a random variable (either **discrete** or **continuous**) $X <= \textbf{certain value x} $
+ It is defined as:
  $$ F(x) = P(X \leq x) $$
  where $F(x)$ is the CDF of the random variable $X$.
+ **Example**: What's the probability that time between arrivals of emails is less than 5 minutes?

</div>

## Distribution Examples

### Binomial Distribution 🪙
<div class="group">
 <div class="text"> 

+ It describes the number of successes in a fixed number of indipendent trials/experiments, where each trial has two possible outcomes (success or failure).
+ It is defined as:
  $$ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} $$
  where:
  - $\binom{n}{k}$ is the [binomial coefficient](https://en.wikipedia.org/wiki/Binomial_coefficient) (number of ways to choose $k$ successes from $n$ trials)
    
  - $n$ is the number of trials
  - $k$ is the number of successes
  - $p$ is the probability of success in each trial (0.5 for a fair coin)
+ **Example**: Flipping a coin 10 times and counting how many times it lands on heads (success).

</div>
<div class="images">
        <img src="../images/distributions/binom_distr.png" alt="Binomial Distribution"> 
    </div>
</div>

### Poisson Distribution 📧
<div class="group">
 <div class="text"> 

+ It describes the number of events that occur within a fixed interval of time or space, given that events happen independently and at a constant average rate.
+ It is defined as:
  $$ P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!} $$
  where:
  - $\lambda$ is the average rate of events in the interval
  - $k$ is the number of events
  - $!$ is the [factorial function](https://en.wikipedia.org/wiki/Factorial)
+ **Example**: The number of emails received in an hour, where the average rate is 5 emails per hour.


</div>
<div class="images">
        <img src="../images/distributions/poisson_distr.png" alt="Poissonian Distribution"> 
    </div>
</div>

### Uniform Distribution 🎲
<div class="group">
 <div class="text"> 

+ It describes a situation where all outcomes are equally likely within a range.
+ It is defined as:
  $$ P(X = x) = \frac{1}{b - a} $$
  where:
  - $a$ is the minimum value
  - $b$ is the maximum value
  - $x$ is any value between $a$ and $b$

+ **Example**: Rolling a fair die, where each face (1 to 6) has an equal probability of 1/6.

</div>
<div class="images">
        <img src="../images/distributions/uniform_distr.png" alt="Uniform Distribution"> 
    </div>
</div>

### Normal Distribution 📈
<div class="group">
 <div class="text"> 

+ It is a bell-shaped curve where most of data points cluster around the mean, with fewer points as you move away from the mean.
+ It is defined as:
  $$ f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}} $$
  where:
  - $\mu$ is the mean (average)
  - $\sigma$ is the standard deviation (spread of the data)
+ **Example**: Heights of people in a population, IQ scores, or errors.
+ **Applications**: foundation for many statistical methods, including hypothesis testing and confidence intervals.


</div>

### What is a Confidence Interval?
<div group>
 <div class="text">
<div style="margin: 40px auto; text-align: center; max-width: 700px; font-size: 1.4em;">
  <span style="color:violet">Confidence Interval (CI)</span> is a 
  <span style="color:violet">range of values</span> constructed from a sample data so that we can be reasonably sure (usually 95% <span style="color:violet">confidence level</span>) that the <span style="color:red">true population parameter</span> lies somewhere in that range.
</div>
</div>

<div class="images">
    <img src="../images/distributions/doubt_face.png" alt="confidence_intervals">
</div>
</div>



notes:
help us to estimate an unknown population parameter (e.g., mean, proportion) based on sample data -- while accounting for uncertainty.

### Example: Height

Let's say you're estimating the average height of adult women in a city based on a sample, and you compute:
<div style="text-align: left; max-width: 800px; font-size: 1.1em;">
CI = [162 cm, 168 cm] with 95% confidence
</div>

<div class="cell-class">

We are 95% confident that the true average height of all adult women in that city is between 162 cm and 168 cm.
</div>

<div style="display: flex; justify-content: space-between; gap: 20px; max-width: 100%; margin: 40px auto; flex-wrap: wrap;">
    
<div style="flex: 1; min-width: 300px; padding: 20px; border: 1px solid #ddd; border-radius: 8px; background: #e6ffe6; font-size: 0.9em;">
      <h3 style="margin-top: 0;">✅ What it does mean:</h3>
      <p>If we repeated this sampling process many times, and built a 95% CI from each sample:</p>
      <p>About 95% of those intervals would contain the true population parameter.</p>
      <p>The interval reflects the precision of our estimate — wider means more uncertainty; narrower means more precise.</p>
</div>
    
<div style="flex: 1; min-width: 300px; padding: 20px; border: 1px solid #ddd; border-radius: 8px; background: #ffe6e6; font-size: 0.9em;">
     <h3 style="margin-top: 0;">❌ What it does NOT mean:</h3>
     <p>It does not mean there’s a 95% chance the true value is inside this specific interval — the true value is fixed; the interval is random.</p>
     <p>It’s about long-run performance of the method, not a probability about this one interval.</p>
</div>
</div>
