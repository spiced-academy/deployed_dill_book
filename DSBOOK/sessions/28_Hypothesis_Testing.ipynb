Notes:
In order to formulate a hypothesis: 
1. A research question must first be defined (from literature review).
2. A precisely formulated hypothesis about the population can then be derived from the research question

Before starting any research project, it's essential:
- to define what you're trying to find out. (scope of the research)
- This usually begins with a research question, 
- which may lead to forming a hypothesis.

__Reasearch Question__: 
- Starting point for investigation or exploration 
- Broader in scope – asks if there’s an effect or relationship
- Not always directly testable without refinement

__Hypothesis__:
- derived from the research question
- Specific, testable prediction about the relationship or effect
- Narrower in focus – what you expect to find

So in practice, you often start with a research question and then refine it into one or more hypotheses that can be tested with data.

Notes:

Now let’s take a closer look at what a hypothesis really is and how it works in hypothesis testing.

Hypotheses are not simple statements; 
1. they are formulated in such a way that they can be tested with collected data in the course of the research process.


When forming a hypothesis, you want to consider three key questions:
1. __What variable will change__?: which variable are involved: For example, the price of a product, customer behavior, or __house prices__.
2. __How will it change__?: Will it __increase__, __decrease__, or shift in some way?
3. __What will cause the change?__ – This identifies your independent variable, like a new marketing campaign or location near water.

In hypothesis testing, we always set up two possible mutually exclusive explanations:
1. **Null Hypothesis (H₀)**: This is the default position that there is no effect or change. Any observed difference is due to random chance.
2. **Alternative Hypothesis (H₁)**: This is what you are trying to test. It states that there is a real change or effect caused by the variable of interest.
- We try to gather evidence to reject the null hypothesis in favor of the alternative hypothesis.

### Purpose and Limitations of Hypothesis Testing
<div style="max-width: 5000px; margin: 0; padding: 20px; display: flex; gap: 20px; align-items: center; flex-wrap: wrap; border: 1px solid #ddd; border-radius: 8px; background: #f9f9f9;">

<!-- Left side (Text) -->
<div style="flex: 1; min-width: 280px; max-width: 900px; font-size: 0.9em;">
<ul style="margin: 0; padding-left: 20px; list-style: none;">
<li>✅ A hypothesis test is used to test an assumption about a population using information from a sample.</li>
<li>🎯 The goal is to determine whether there is enough evidence to reject the <strong>null hypothesis</strong> or retain it.</li>
<li>⚠️ It <strong>never rejects</strong> the null hypothesis with <strong>absolute certainty</strong>.</li>
<li>🔄 There is always a probability of rejecting the null hypothesis even if it is actually true.</li>
<li>📏 This probability threshold is called the <strong>level of significance α</strong> (commonly 5%).</li>
<li>📊 If the <strong>p-value</strong> is less than α, we reject the null hypothesis in favor of the alternative hypothesis.</li>
</ul>

</ul>

### Two-Tailed (Non-Directional) Hypothesis Test
<div style="max-width: 5000px; margin: 0; display: flex; gap: 20px; flex-wrap: wrap; align-items: flex-start;">

<!-- Left: Text -->
<div style="flex: 0 0 auto; max-width: 800px; min-width: 750px; border: 1px solid #ddd; border-radius: 8px; background: #f9f9f9; padding: 20px; font-size: 0.95em; line-height: 1.5;">
<p><strong>Purpose</strong>:<br>
Test for a difference in <strong>both directions</strong>.
</p>

<p><strong>How it works</strong>:</p>
<ul style="margin: 0 0 1em 20px; padding: 0;">
<li>If your sample result falls in either critical region, you reject the null hypothesis and accept the alternative hypothesis.</li>
<li>If it does not, you don't reject the null hypothesis (no evidence to support the alternative hypothesis).</li>
</ul>

<p><strong>Example</strong>:<br>
A real estate company claims homes near a waterfall are either more or less expensive.
</p>

<p><strong>Hypotheses</strong>:</p>
<ul style="margin: 0 0 1em 20px; padding: 0;">
<li>Null Hypothesis (<strong>H₀</strong>): μ₁ = μ₂ (True mean price near waterfall is equal to that of homes not near)</li>
<li>Alternative Hypothesis (<strong>H₁</strong>): μ₁ ≠ μ₂ (True mean price near waterfall is different from that of homes not near)</li>
</ul>

Notes:
As the __alternative hypothesis does not predict a specific direction__ (μ₁ ≠ μ₂),
+ you perform a __two-tailed test__ on H0.
+ Specifically, a __two-tailed test__ based on the __standardized mean difference__ (e.g., a t-statistic or z-statistic).
+ alpha is significance level describes the probability of rejecting H0 when it is true.
+ The alpha level is typically set at 0.05, which means you are willing to accept a 5% chance of making a Type I error (rejecting H0 when it is true).

The plot you're about to see shows a standard normal distribution — the familiar bell curve — and highlights the rejection regions for a two-tailed test.
For test we assume H0 is true, so we expect the sample mean difference to be equal to 0.
The bell curve (standard normal distribution) represents the range of possible outcomes we’d expect if the null hypothesis is true.
The critical values (Z = ±1.96) marks the points beyond which we say:
* "This result is unlikely under H₀."
* The shaded areas show the __rejection regions__: if your test statistic falls here, you conclude the __sample mean__ is too extreme to be explained by H₀.

* The sample mean difference can be less than 0, equal to 0, or even greater than 0, just by chance if H₀ is true.
But How far can the sample mean difference deviate from 0 before we say:
* "This is too extreme to be explained by H₀"?
* If Z falls within the rejection regions, we reject H₀. otherwise, we fail to reject H₀.

### Understanding the p-value
<div style="max-width: 5000px; margin: 0; display: flex; gap: 20px; flex-wrap: wrap; align-items: flex-start;">

<!-- Left: Text -->
<div style="flex: 0 0 auto; max-width: 800px; min-width: 750px; border: 1px solid #ddd; border-radius: 8px; background: #f9f9f9; padding: 20px; font-size: 0.95em; line-height: 1.6;">
<p><strong>Purpose</strong>:<br>
Used to decide whether to <strong>reject</strong> or <strong>retain</strong> the null hypothesis.
</p>

<p><strong>Definition</strong>:<br>
The p-value is the probability (assuming the null hypothesis is true) of obtaining a sample result
at least as extreme as the one observed.
</p>

<p><strong>Decision rule</strong>:</p>
<ul style="margin: 0 0 1em 20px; padding: 0;">
<li>If <strong>p-value ≤ α</strong> (e.g., 0.05), reject H₀.</li>
<li>If <strong>p-value > α</strong>, do not reject H₀.</li>
</ul>

<p><strong>Notes</strong>:</p>
<ul style="margin: 0 0 0.5em 20px; padding: 0;">
<li>Smaller p-values give stronger evidence against H₀.</li>
<li>The p-value is <em>not</em> the probability that H₀ is true.</li>
</ul>
</div>

## A/B Testing


### 🆎 Introduction to A/B Testing
<div style="max-width: 900px; margin: 0; padding: 30px; border-radius: 8px; background: #f9f9f9; font-family: Arial, sans-serif; line-height: 1.5;">

<h4 style="margin-top: 0; color: #333;">What is A/B Testing?</h4>
<p style="margin-bottom: 1em;">
A/B Testing is an <strong>experiment</strong> where you compare two versions (A and B) of something to see which performs better. 

<h4 style="margin-top: 20px; color: #333;">Why We Use It</h4>
<ul style="margin-bottom: 1em; padding-left: 20px;">
<li>To make <strong>data-driven decisions</strong> instead of relying on guesswork.</li>
<li>To improve performance and user experience.</li>
<li>To validate design, content, or product changes before full rollout.</li>
</ul>
</div>



### How A/B Testing Works
<div style="max-width: 900px; margin: 0; padding: 30px; border-radius: 8px; background: #f9f9f9; font-family: Arial, sans-serif; line-height: 1.5;">

<h4 style="margin-top: 20px; color: #333;">How it Works</h4>
<ul style="margin-bottom: 1em; padding-left: 20px;">
<li>Randomly split your audience into two groups.</li>
<li>Show group A one version and group B another version.</li>
<li>Measure the outcome (click rate, purchase rate, etc.).</li>
<li>Use a statistical test (often a <em>two-sample proportion test</em> or <em>t-test</em>) to see if the difference is significant.</li>
</ul>

<div style="display: flex; gap: 20px; margin-top: 20px; flex-wrap: wrap;">

<!-- Example -->
<div style="flex: 1; min-width: 250px; background: #e8f4fd; padding: 15px; border-radius: 6px;">
<h5 style="margin-top: 0;">Example</h5>
<p>💻 A company tests two landing page designs:</p>
<ul style="padding-left: 20px; margin: 0;">
<li><strong>Version A</strong>: Without Chatbot</li>
<li><strong>Version B</strong>: With Chatbot</li>
</ul>
<p style="margin-top: 0.5em;">Measure which version leads to more sign-ups.</p>
</div>

<!-- Diagram -->
<div style="flex: 1; min-width: 250px; background: #fff; padding: 15px; border-radius: 6px; text-align: center; border: 1px solid #ddd;">
    <img src="../images/hypothesis_testing/random_split_experiment.png" alt="A/B Testing Diagram" style="max-width: 100%; height: auto; border-radius: 4px;">
</div>
</div>

<h4 style="margin-top: 20px; color: #333;">Key Points</h4>
<ul style="padding-left: 20px; margin: 0;">
<li><strong>H₀</strong>: No difference between A and B.</li>
<li><strong>H₁</strong>: One version performs better.</li>
<li>Significance level (α).</li>
</ul>
</div>


### Why Random Assignment 
<div style="max-width: 900px; margin: 0; padding: 30px; border-radius: 8px; background: #f9f9f9; font-family: Arial, sans-serif; line-height: 1.5;">

It ensures that the groups are <strong>comparable</strong> and any differences in outcomes are likely due to the change you made — not to pre-existing differences.



<h5 style="margin-top: 0;">Why It Matters</h5>
<ul style="padding-left: 20px; margin: 0;">
<li><strong>Removes Selection Bias</strong> – Ensures both groups start out similar.</li>
<li><strong>Balances Confounding Variables</strong> – Distributes known and unknown factors evenly.</li>
</ul>
<p style="margin-top: 1em; font-style: italic;">
Without random assignment, differences may be due to who’s in the groups — not the change you made.
</p>
</div>


### Interpreting A/B Test Results

<div style="max-width: 1100px; margin: 0 auto; padding: 30px; border-radius: 8px; background: #f9f9f9; font-family: Arial, sans-serif; line-height: 1.5;">

<h4 style="margin-top: 0; color: #333;">Key Outputs from an A/B Test</h4>
<ul style="margin-bottom: 1em; padding-left: 20px;">
<li><strong>p-value</strong> – Probability the observed difference happened by chance.</li>
<li><strong>Confidence Interval (CI)</strong> – Range of likely true differences.</li>
<li><strong>Effect Size</strong> – Magnitude of the difference (practical significance).</li>
</ul>

Statistical significance (<strong>p-value</strong>) is not enough — always check the <strong>effect size</strong> and the <strong>confidence interval</strong>.
</p>

### Conclusion - Hypothesis Testing & A/B Testing
<div style="max-width: 900px; margin: 0; padding: 30px; border-radius: 8px; background: #f4f9ff; font-family: Arial, sans-serif; line-height: 1.5;">

  <h4 style="margin-top: 0; color: #004c99;">Key Takeaways</h4>
  <ul style="padding-left: 20px; margin-bottom: 1em;">
    <li>Both <strong>Hypothesis Testing</strong> and <strong>A/B Testing</strong> help make data-driven decisions rather than relying on guesswork.</li>
    <li>Every test starts with a clear <strong>null hypothesis (H₀)</strong> and <strong>alternative hypothesis (H₁)</strong>.</li>
    <li>The <strong>p-value</strong> tells us how likely it is to observe the data (or more extreme) if H₀ were true — but it’s not absolute proof.</li>
    <li>Statistical significance doesn’t always mean <em>practical significance</em> — always check effect sizes and context.</li>
  </ul>
</div>
